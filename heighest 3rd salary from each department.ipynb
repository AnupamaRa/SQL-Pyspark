```sql
---DROP TABLE employeN;
---DROP TABLE department;
CREATE TABLE employeN (
    emp_name VARCHAR(50),
    id INT,
    dept_id INT,
    salary INT
);

CREATE TABLE department (
    dept_id INT,
    department VARCHAR(50)
);

INSERT INTO employeN (emp_name, id, dept_id, salary) VALUES
('Anu', 1, 10, 50000),
('Raj', 2, 10, 60000),
('Sim', 3, 10, 55000),
('Rani', 4, 10, 48000),
('Pooja', 5, 20, 70000),
('Amit', 6, 20, 65000),
('Ravi', 7, 20, 62000),
('Neha', 8, 20, 58000);

INSERT INTO department (dept_id, department) VALUES
(10, 'IT'),
(20, 'HR');


SELECT 
    e.dept_id, 
    d.department, 
    e.emp_name, 
    e.salary
FROM (
    SELECT 
        emp_name, 
        dept_id, 
        salary,
        ROW_NUMBER() OVER (PARTITION BY dept_id ORDER BY salary DESC) AS rank
    FROM employeN
) e
JOIN department d
    ON e.dept_id = d.dept_id
WHERE e.rank = 3;

```
